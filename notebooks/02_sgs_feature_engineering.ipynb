# 02 — S_sgs 特征工程

检查无量纲特征与残余目标的相关性。

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
ROOT = Path('..').resolve()
df = pd.read_csv(ROOT / 'data_proc/sgs_training_data.csv')
dim = ['Fr','Re','Pe','Da','k_tau_h','h_over_W','DOC_over_DIC']
cols = [c for c in dim if c in df.columns] + ['S_sgs_residual_mol_m2d']
sns.pairplot(df[cols].sample(min(500, len(df))), diag_kind='kde')
plt.savefig(ROOT / 'results/figures/sgs_feature_pairplot.png', dpi=120)
plt.show()
```
